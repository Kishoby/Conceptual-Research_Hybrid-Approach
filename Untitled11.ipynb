```python
import pandas as pd
import numpy as np

from river import metrics, compose, preprocessing
```

```python
multi_target_df = pd.read_csv("/content/drive/MyDrive/Research v2/Research 18.03.2026/Multi Target/cleaned_multi_target_dataset (1).csv")

print("Dataset shape:", multi_target_df.shape)
multi_target_df.head()
```

```python
targets = [
    "air_quality_PM2.5",
    "temperature_celsius",
    "humidity"
]

features = [col for col in multi_target_df.columns if col not in targets]

print("Targets:", targets)
print("Number of features:", len(features))
print(features)
```

```python
for col in targets:
    multi_target_df[f"{col}_lag1"] = multi_target_df[col].shift(1)
    multi_target_df[f"{col}_lag2"] = multi_target_df[col].shift(2)

multi_target_df = multi_target_df.dropna().reset_index(drop=True)

print("Shape after adding target lags:", multi_target_df.shape)
multi_target_df.head()
```

```python
features = [col for col in multi_target_df.columns if col not in targets]

print("Updated feature count:", len(features))
print(features)
```

```python
split_index = int(0.8 * len(multi_target_df))

train_df = multi_target_df.iloc[:split_index].copy()
test_df = multi_target_df.iloc[split_index:].copy()

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)
```

```python
n_train = len(train_df)

iter1 = train_df.iloc[:int(0.33 * n_train)].copy()
iter2 = train_df.iloc[int(0.33 * n_train):int(0.66 * n_train)].copy()
iter3 = train_df.iloc[int(0.66 * n_train):].copy()

print("Iteration 1 shape:", iter1.shape)
print("Iteration 2 shape:", iter2.shape)
print("Iteration 3 shape:", iter3.shape)
```

```python
def build_hybrid_model():
    try:
        from river import forest
        model = forest.ARFRegressor(
            n_models=20,
            max_features="sqrt",
            lambda_value=6,
            seed=42
        )
        model_name = "River ARF Hybrid"
    except Exception:
        from river import tree
        model = compose.Pipeline(
            preprocessing.StandardScaler(),
            tree.HoeffdingAdaptiveTreeRegressor(
                grace_period=50,
                delta=1e-5,
                leaf_prediction="adaptive"
            )
        )
        model_name = "River HAT Hybrid"

    return model, model_name
```

```python
target_models = {}
model_name = None

for target in targets:
    model, model_name = build_hybrid_model()
    target_models[target] = model

print("Using hybrid model:", model_name)
print("Models created for targets:", list(target_models.keys()))
```

```python
def run_multi_target_hybrid_iteration(data, target_models, iteration_name, features, targets, warmup=False):
    metric_store = {
        target: {
            "mse": metrics.MSE(),
            "rmse": metrics.RMSE(),
            "mae": metrics.MAE(),
            "r2": metrics.R2()
        }
        for target in targets
    }

    first_part = int(0.1 * len(data)) if warmup else 0

    for i, (_, row) in enumerate(data.iterrows()):
        x = row[features].to_dict()

        for target in targets:
            y = row[target]

            # warmup for first part of iteration 1
            if i < first_part:
                target_models[target].learn_one(x, y)
                continue

            y_pred = target_models[target].predict_one(x)
            if y_pred is None:
                y_pred = 0.0

            metric_store[target]["mse"].update(y, y_pred)
            metric_store[target]["rmse"].update(y, y_pred)
            metric_store[target]["mae"].update(y, y_pred)
            metric_store[target]["r2"].update(y, y_pred)

            target_models[target].learn_one(x, y)

    rows = []
    for target in targets:
        mse = metric_store[target]["mse"].get()
        rmse = metric_store[target]["rmse"].get()
        mae = metric_store[target]["mae"].get()
        r2 = metric_store[target]["r2"].get()
        acc = r2 * 100

        rows.append([
            iteration_name,
            model_name,
            target,
            mse,
            rmse,
            mae,
            r2,
            acc
        ])

    results_df = pd.DataFrame(
        rows,
        columns=["Iteration", "Model", "Target", "MSE", "RMSE", "MAE", "R2", "Accuracy (%)"]
    ).round(3)

    print(f"\n{iteration_name} Results")
    display(results_df)

    return target_models, results_df
```

```python
target_models, hybrid_iter1_df = run_multi_target_hybrid_iteration(
    iter1,
    target_models,
    "Iteration 1",
    features,
    targets,
    warmup=True
)
```

```python
target_models, hybrid_iter2_df = run_multi_target_hybrid_iteration(
    iter2,
    target_models,
    "Iteration 2",
    features,
    targets,
    warmup=False
)
```

```python
target_models, hybrid_iter3_df = run_multi_target_hybrid_iteration(
    iter3,
    target_models,
    "Iteration 3",
    features,
    targets,
    warmup=False
)
```

```python
hybrid_iteration_results = pd.concat([
    hybrid_iter1_df,
    hybrid_iter2_df,
    hybrid_iter3_df
], ignore_index=True)

print("Hybrid Multi-target - All Training Iteration Results")
display(hybrid_iteration_results)
```

```python
def evaluate_multi_target_hybrid_test(data, target_models, features, targets, model_name):
    metric_store = {
        target: {
            "mse": metrics.MSE(),
            "rmse": metrics.RMSE(),
            "mae": metrics.MAE(),
            "r2": metrics.R2()
        }
        for target in targets
    }

    for _, row in data.iterrows():
        x = row[features].to_dict()

        for target in targets:
            y = row[target]

            y_pred = target_models[target].predict_one(x)
            if y_pred is None:
                y_pred = 0.0

            metric_store[target]["mse"].update(y, y_pred)
            metric_store[target]["rmse"].update(y, y_pred)
            metric_store[target]["mae"].update(y, y_pred)
            metric_store[target]["r2"].update(y, y_pred)

    rows = []
    for target in targets:
        mse = metric_store[target]["mse"].get()
        rmse = metric_store[target]["rmse"].get()
        mae = metric_store[target]["mae"].get()
        r2 = metric_store[target]["r2"].get()
        acc = r2 * 100

        rows.append([
            model_name,
            target,
            mse,
            rmse,
            mae,
            r2,
            acc
        ])

    test_df = pd.DataFrame(
        rows,
        columns=["Model", "Target", "MSE", "RMSE", "MAE", "R2", "Accuracy (%)"]
    ).round(3)

    return test_df
```

```python
hybrid_test_results = evaluate_multi_target_hybrid_test(
    test_df,
    target_models,
    features,
    targets,
    model_name
)

print("Hybrid Multi-target - Final Test Results")
display(hybrid_test_results)
```

```python
hybrid_rmse_iter_table = hybrid_iteration_results.pivot(
    index="Iteration",
    columns="Target",
    values="RMSE"
).round(3)

print("Hybrid Multi-target - Training RMSE Table")
display(hybrid_rmse_iter_table)
```
